```python
from PIL import Image
from pathlib import Path
```

```python
FULL_DIR = Path('outputs/generate_hvm_obj/full')
IMAGE_FILE = 'stim0290.png'
PANEL_LABELS = ['GT', 'Control', 'Text', 'Neural pred', 'GT emb', 'Obj crop']
LABEL_ROWS = 36
```

```python
img = Image.open(FULL_DIR / IMAGE_FILE)
w, h = img.size
panel_w = w // len(PANEL_LABELS)
panels = [img.crop((i * panel_w, LABEL_ROWS, (i + 1) * panel_w, h)) for i in range(len(PANEL_LABELS))]
```

```python
for panel in panels:
    display(panel)
```
